```python
output_v = glob.glob(os.path.join(os.getcwd(), 'Output_Videos', '*'))

for file in output_v:
    basename = os.path.basename(file)
    #basename = basename.replace('.avi', '')
    if basename not in tokeep:
        os.remove(file)

```
